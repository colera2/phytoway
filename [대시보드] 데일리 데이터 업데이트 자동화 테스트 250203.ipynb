```python
from sqlalchemy import create_engine, text
import urllib.parse
import os


# 1. 연결에 필요한 정보 설정 (실제 값으로 변경)
username = "dev"
host = "db1-dev.postgres.database.azure.com"
port = "5432"
database = "dainDB"
password = "@Vkdlxhdnpdl0"  

# 2. 비밀번호 인코딩하기
encoded_password = urllib.parse.quote(password)
print("인코딩된 비밀번호:", encoded_password)

# 3. DSN 문자열 작성하기
connection_string = f"postgresql+psycopg2://{username}:{encoded_password}@{host}:{port}/{database}"
print("연결 문자열:", connection_string)

# 4. 엔진 생성 및 연결 테스트
engine = create_engine(connection_string, connect_args={"options": "-c client_encoding=UTF8"})

try:
    with engine.connect() as connection:
        result = connection.execute(text("SELECT 'Hello World' AS greeting;"))
        print("연결 성공! 결과:", result.fetchone()[0])
except Exception as e:
    print("오류 발생:", e)
```
